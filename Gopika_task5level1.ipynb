```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

try:
    from textblob import TextBlob
except ImportError:
    import subprocess
    import sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "textblob"])
    from textblob import TextBlob

# Load dataset
file_path = Path("review.csv")
if not file_path.exists():
    raise FileNotFoundError(f"The file '{file_path.name}' was not found in: {Path.cwd()}")

df = pd.read_csv(file_path)

if "Review" not in df.columns:
    raise ValueError("The dataset must contain a 'Review' column.")

print("SENTIMENT ANALYSIS")
print("=" * 40)

# Display first records
print("\nFirst 5 Review:")
print(df.head())

# Dataset information
print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Remove missing reviews
df = df.dropna(subset=["Review"]).copy().reset_index(drop=True)

# Sentiment analysis function
def get_sentiment(text):
    polarity = TextBlob(str(text)).sentiment.polarity

    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"


# Calculate polarity and sentiment
df["Polarity"] = df["Review"].apply(lambda x: TextBlob(str(x)).sentiment.polarity)
df["Sentiment"] = df["Review"].apply(get_sentiment)

# Display results
print("\nSentiment Analysis Results:")
print(df[["Review", "Polarity", "Sentiment"]].head(10))

# Count sentiments
print("\nSentiment Counts:")
sentiment_counts = df["Sentiment"].value_counts()
print(sentiment_counts)

# Sentiment visualization
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Sentiment")
plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()

# Polarity distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Polarity", bins=10, kde=True)
plt.title("Sentiment Polarity Distribution")
plt.xlabel("Polarity")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()

# Save results
df.to_csv("sentiment_analysis_results.csv", index=False)
print("\nSentiment analysis results saved successfully!")

# Conclusion
print("""
CONCLUSION

Sentiment analysis was performed on customer reviews
using TextBlob.

The reviews were classified into Positive, Negative,
and Neutral sentiments based on their polarity scores.

This analysis helps understand customer opinions
and overall customer satisfaction.
""")
```
